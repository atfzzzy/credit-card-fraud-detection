Under-sampling


*   Build a sample dataset containing similar distribution of normal transactions and fraudulent transactions
*  No. of fraudulent transactions=492



```python
legit_sample = legit.sample(n=492, random_state=42)
```

```python
new_dataset=pd.concat([legit_sample,fraud], axis=0)
```

```python
new_dataset.head()
```

```python
new_dataset.tail()
```

```python
new_dataset['Class'].value_counts()
```

```python
new_dataset.groupby('Class').mean()
```

```python
X=new_dataset.drop(columns="Class",axis=1)
Y=new_dataset["Class"]
```

```python
print(X)
```

```python
print(Y)
```

```python
# Splitting the data into Training and Testing data

X_train, X_test, Y_train, Y_test = train_test_split(
    X,
    Y,
    test_size=0.2,
    stratify=Y,
    random_state=2
)
```

```python
print(X.shape, X_train.shape, X_test.shape)
```

Model Training



Logistic Regression


```python
# Scaling the features
scaler = StandardScaler()

X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)
```

```python
# Training the Logistic Regression Model
model = LogisticRegression(max_iter=2000)

model.fit(X_train_scaled, Y_train)
print("Model training completed.")
```

```python
# Accuracy on training data
X_train_prediction = model.predict(X_train_scaled)

training_data_accuracy = accuracy_score(
    Y_train,
    X_train_prediction
)

print('Accuracy on Training data : ', training_data_accuracy)
```

```python
# Accuracy on test data
X_test_prediction = model.predict(X_test_scaled)

test_data_accuracy = accuracy_score(
    Y_test,
    X_test_prediction
)

print('Accuracy score on Test Data : ', test_data_accuracy)
```

```python
## Classification Report
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(Y_test, X_test_prediction))
```

```python
#Confusion Matrix
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

cm = confusion_matrix(Y_test, X_test_prediction)

sns.heatmap(cm, annot=True, fmt='d', cmap='Blues')

plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.title('Confusion Matrix')

plt.show()
```

```python
#ROC-AUC Score
from sklearn.metrics import roc_auc_score

X_test_probability = model.predict_proba(X_test_scaled)[:, 1]

roc_auc = roc_auc_score(Y_test, X_test_probability)

print("ROC-AUC Score:", roc_auc)
```
